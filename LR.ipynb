Question 1 : What is Simple Linear Regression (SLR)? Explain its purpose.

Simple Linear Regression SLR is a statistical method used to model the relationship between one independent variable and one dependent variable. It helps explain how changes in one variable affect another by fitting a straight line to observed data.

The general form of Simple Linear Regression is:

Y = a + bX

Where
Y is the dependent variable
X is the independent variable
a is the intercept which represents the value of Y when X is zero
b is the slope which represents the change in Y for a one unit change in X

The main purpose of Simple Linear Regression is to:

Understand the relationship between two variables
It shows whether the relationship is positive, negative, or neutral and how strong that relationship is.

Make predictions
Once the regression line is calculated, it can be used to predict the value of the dependent variable based on a given value of the independent variable.

Measure the impact of one variable on another
The slope helps quantify how much the dependent variable changes when the independent variable changes.

Support decision making
Businesses, researchers, and analysts use SLR to analyze trends and make data driven decisions.

In simple terms, Simple Linear Regression helps in drawing a best fit straight line through data points to study and predict relationships between two variables.

Question 2: What are the key assumptions of Simple Linear Regression?

The key assumptions of Simple Linear Regression are the conditions that must be satisfied for the model to give reliable results.

Linearity
There must be a straight line relationship between the independent variable X and the dependent variable Y. This means changes in X lead to proportional changes in Y.

Independence of errors
The residuals or errors should be independent of each other. One observation should not influence another.

Homoscedasticity
The variance of errors should remain constant for all values of X. In other words, the spread of residuals should be roughly the same across the regression line.

Normality of errors
The residuals should be normally distributed, especially important for hypothesis testing and confidence intervals.

No significant outliers
Extreme values can strongly affect the regression line and distort results, so data should be free from influential outliers.

X is measured without error
The independent variable is assumed to be measured accurately.

Question 3: Write the mathematical equation for a simple linear regression model and
explain each term.

The mathematical equation of a Simple Linear Regression model is:

Y = a + bX + e

Explanation of each term:

Y
Y is the dependent variable. It represents the outcome or response we want to predict or explain.

X
X is the independent variable. It is the predictor or input variable used to estimate Y.

a
a is the intercept. It is the value of Y when X equals zero. It shows where the regression line crosses the Y axis.

b
b is the slope or regression coefficient. It represents how much Y changes for every one unit increase in X. It indicates the direction and strength of the relationship between X and Y.

e
e is the error term. It represents the difference between the actual observed value of Y and the value predicted by the model. It accounts for all other factors not included in the regression equation.

In summary, this equation describes a straight line that best fits the data and is used to understand relationships and make predictions.

Question 5: What is the method of least squares in linear regression?

The method of least squares in linear regression is a mathematical technique used to find the best fitting straight line through a set of data points.

Its main idea is to minimize the total error between the actual observed values and the values predicted by the regression line.

In Simple Linear Regression, this is done by minimizing the sum of the squared residuals, where a residual is the difference between an observed value and its predicted value.

Residual = Actual value minus Predicted value

The method chooses the values of a (intercept) and b (slope) such that:

The sum of squared residuals is as small as possible.

Purpose of the method of least squares:

To obtain the most accurate regression line

To reduce prediction errors

To provide unbiased and consistent estimates of regression parameters

By squaring the residuals, positive and negative errors do not cancel each other out, and larger errors receive more penalty. This ensures the regression line represents the overall trend of the data as closely as possible.

In simple words, the method of least squares finds the line that best represents the data by making the total squared prediction error minimum.

Question 6: What is Logistic Regression? How does it differ from Linear Regression?

Logistic Regression is a statistical and machine learning method used to predict outcomes that are categorical, most commonly binary such as yes or no, pass or fail, or 0 or 1. Instead of predicting exact numerical values, it estimates the probability that an observation belongs to a particular class.

Logistic Regression uses an S shaped curve called the logistic function to map predicted values between 0 and 1. Based on this probability, the model assigns the final class.

Purpose of Logistic Regression:

To classify data into categories

To estimate the probability of an event occurring

To support decision making in classification problems such as disease detection, spam filtering, or customer churn prediction

Difference between Logistic Regression and Linear Regression:

Type of output
Linear Regression predicts continuous numerical values such as price or temperature.
Logistic Regression predicts probabilities and categorical outcomes such as true or false.

Nature of relationship
Linear Regression fits a straight line to the data.
Logistic Regression fits an S shaped curve.

Range of predictions
Linear Regression can produce any real value from negative to positive infinity.
Logistic Regression always produces values between 0 and 1.

Typical use cases
Linear Regression is used for forecasting quantities like sales or house prices.
Logistic Regression is used for classification tasks like predicting whether a customer will buy a product.

In summary, Linear Regression is used for predicting continuous values, while Logistic Regression is used for predicting categories by estimating probabilities.

Question 7: Name and briefly describe three common evaluation metrics for regression
models.

Three common evaluation metrics for regression models are:

Mean Absolute Error MAE
MAE measures the average of the absolute differences between actual values and predicted values.

It tells us how far predictions are from real values on average.

Smaller MAE indicates better model performance.

It is easy to understand because errors are in the same units as the target variable.

Mean Squared Error MSE
MSE calculates the average of the squared differences between actual and predicted values.

Because errors are squared, larger mistakes are penalized more heavily.

A lower MSE means the model fits the data better.

It is widely used in optimization during model training.

Root Mean Squared Error RMSE
RMSE is the square root of MSE.

It represents the standard deviation of prediction errors and shows how concentrated the data is around the best fit line.

Like MAE, RMSE is expressed in the same units as the dependent variable, but it gives more weight to large errors.

Lower RMSE indicates better accuracy.

In summary, these metrics help measure how close model predictions are to actual values, with lower values showing better regression performance.

Question 8: What is the purpose of the R-squared metric in regression analysis?

The purpose of the R squared metric in regression analysis is to measure how well the independent variable explains the variation in the dependent variable.

R squared is also called the coefficient of determination. Its value ranges from 0 to 1.

Meaning of R squared values:

An R squared value of 0 means the model explains none of the variation in the output.
An R squared value of 1 means the model explains all the variation in the output.
Values between 0 and 1 show partial explanation.

Purpose of R squared:

To evaluate model performance
It shows how well the regression model fits the data.

To measure explained variance
It indicates the percentage of variability in the dependent variable that is explained by the independent variable.

To compare models
Higher R squared values generally indicate a better fitting model when comparing similar regression models.

For example, an R squared value of 0.80 means that 80 percent of the variation in the dependent variable is explained by the model, while the remaining 20 percent is due to other factors or random error.

In simple terms, R squared tells us how much of the outcome can be explained by the predictor and helps judge the overall quality of a regression model.



```python
#Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.
#(Include your Python code and output in the code box below.)

import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data (X = hours studied, Y = marks scored)
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2, 4, 6, 8, 10])

# Create and fit the model
model = LinearRegression()
model.fit(X, Y)

# Get slope and intercept
slope = model.coef_[0]
intercept = model.intercept_

print("Slope:", slope)
print("Intercept:", intercept)
```

Question 10: How do you interpret the coefficients in a simple linear regression model?

In a Simple Linear Regression model, the coefficients explain how the independent variable affects the dependent variable. The model is written as:

Y = a + bX

Interpretation of the coefficients:

Intercept a
The intercept represents the predicted value of Y when X equals zero.
It shows where the regression line crosses the Y axis.

In real-world terms, it is the baseline value of the dependent variable before any influence from the independent variable.

Slope b
The slope indicates how much Y changes for every one unit increase in X.

If b is positive, Y increases as X increases.
If b is negative, Y decreases as X increases.
If b is zero, X has no effect on Y.

The magnitude of b shows the strength of the relationship between X and Y.

In summary, the intercept gives the starting value of Y, and the slope shows the direction and amount of change in Y caused by X. Together, these coefficients describe the relationship between the two variables and help make predictions.
