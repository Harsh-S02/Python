Question 1: Explain the differences between AI, ML, Deep Learning (DL), and Data
Scienc(DS).

=> Artificial Intelligence (AI), Machine Learning (ML), Deep Learning (DL), and Data Science (DS) are closely related fields, but they are not the same. Each has a different scope and purpose.

Artificial Intelligence (AI) is the broadest concept. It refers to the development of computer systems that can perform tasks that normally require human intelligence. These tasks include reasoning, problem-solving, understanding language, recognizing images, and making decisions. AI can include rule-based systems, expert systems, robotics, natural language processing, and learning-based approaches. In simple terms, AI is the overall goal of making machines “smart.”

Machine Learning (ML) is a subset of AI. It focuses on creating algorithms that allow computers to learn from data and improve their performance without being explicitly programmed for every task. Instead of writing detailed rules, developers provide data, and the system identifies patterns and makes predictions or decisions. Examples include spam detection, recommendation systems, and fraud detection. ML is one of the most important techniques used to achieve AI.

Deep Learning (DL) is a specialized subset of Machine Learning. It uses artificial neural networks inspired by the structure of the human brain. These networks have multiple layers (hence “deep”) that allow them to learn complex patterns from large amounts of data. Deep Learning is especially powerful in tasks like image recognition, speech recognition, and natural language processing. Technologies such as self-driving cars and voice assistants heavily rely on DL models. DL requires large datasets and high computational power compared to traditional ML methods.

Data Science (DS) is a broader field that focuses on extracting insights and knowledge from data. It combines statistics, programming, data analysis, and domain expertise. Data Science involves collecting, cleaning, analyzing, and visualizing data to support decision-making. While Data Science often uses Machine Learning and Deep Learning techniques, its main goal is to understand and interpret data, not necessarily to build intelligent systems. In short, ML and DL are tools that Data Scientists may use as part of their work.

To summarize the relationship: AI is the overall concept of building intelligent systems. ML is a method within AI that allows systems to learn from data. DL is a more advanced type of ML that uses deep neural networks. Data Science is a separate but related field focused on analyzing data and generating insights, often using ML and DL as tools.


Question 2: What are the types of machine learning? Describe each with one
real-world example.

=> There are mainly four types of Machine Learning: Supervised Learning, Unsupervised Learning, Semi-Supervised Learning, and Reinforcement Learning.

1. Supervised Learning

In supervised learning, the model is trained using labeled data, which means each input comes with a known correct output. The algorithm learns the relationship between inputs and outputs and then uses this knowledge to make predictions on new data. This type is commonly used for classification and regression tasks.

Real-world example:
Email spam detection. The system is trained on emails labeled as “spam” or “not spam,” and after learning the patterns, it can classify new incoming emails automatically.

2. Unsupervised Learning

In unsupervised learning, the data has no labels. The algorithm tries to discover hidden patterns or structures in the data on its own. It is mainly used for clustering and association tasks.

Real-world example:
Customer segmentation in marketing. Businesses group customers based on purchasing behavior to identify different customer types, even though no predefined categories are given.

3. Semi-Supervised Learning

Semi-supervised learning uses a small amount of labeled data along with a large amount of unlabeled data. This approach is helpful when labeling data is expensive or time-consuming. The model learns from both labeled and unlabeled samples to improve accuracy.

Real-world example:
Image recognition systems where only some images are labeled (for example, “cat” or “dog”), while thousands of unlabeled images help the model learn visual patterns more effectively.

4. Reinforcement Learning

Reinforcement learning is based on learning through interaction with an environment. The model (called an agent) takes actions and receives rewards or penalties. Over time, it learns the best actions to maximize total reward. There are no labeled answers—learning happens through trial and error.

Real-world example:
Self-driving cars learning how to navigate roads safely by continuously receiving feedback from their driving decisions, such as staying in lane or avoiding obstacles.

Question 3: Define overfitting, underfitting, and the bias-variance tradeoff in machine
learning.


=> In machine learning, overfitting, underfitting, and the bias–variance tradeoff describe how well a model learns from data and how it performs on unseen examples.

Overfitting

Overfitting happens when a model learns the training data too well, including noise and random fluctuations. As a result, it achieves very high accuracy on training data but performs poorly on new (test) data. This usually occurs when the model is too complex or trained for too long. An overfitted model memorizes instead of generalizing.

Example: A model that perfectly predicts past house prices but gives inaccurate results for future houses because it learned irrelevant details from the training set.

Underfitting

Underfitting occurs when a model is too simple to capture the underlying patterns in the data. It performs poorly on both training and test data because it has not learned enough from the dataset. This typically happens when the model is overly basic or trained insufficiently.

Example: Using a straight-line model to represent complex relationships in medical or financial data.

Bias–Variance Tradeoff

The bias–variance tradeoff explains the balance between two sources of error:

Bias is the error caused by overly simple assumptions in the model. High bias leads to underfitting.

Variance is the error caused by the model being too sensitive to small changes in training data. High variance leads to overfitting.

A good machine learning model aims to find the right balance between bias and variance. Reducing bias usually increases variance, and reducing variance usually increases bias. The goal is to minimize both so the model generalizes well to unseen data.


Question 4: What are outliers in a dataset, and list three common techniques for
handling them.

=>Outliers are data points that are significantly different from the rest of the values in a dataset. They may appear unusually high or low compared to most observations. Outliers can occur due to measurement errors, data entry mistakes, or genuine rare events. If not handled properly, they can distort statistical results and negatively affect machine learning models by skewing averages, increasing variance, and misleading predictions.

Three common techniques for handling outliers are:

1. Removal (Trimming)
In this method, extreme values are simply removed from the dataset. This is appropriate when outliers are clearly caused by errors (such as incorrect sensor readings or typing mistakes). However, this should be done carefully to avoid losing important information.

2. Transformation (Scaling or Log Transformation)
Here, mathematical transformations such as logarithmic, square root, or standard scaling are applied to reduce the impact of extreme values. Instead of deleting outliers, this approach compresses large values so they don’t dominate the model.

3. Capping (Winsorization)
In this technique, outliers are replaced with boundary values (for example, replacing everything above the 95th percentile with the 95th-percentile value). This keeps all data points while limiting the influence of extreme values.


Question 5: Explain the process of handling missing values and mention one
imputation technique for numerical and one for categorical data.

=> Handling missing values is an important step in data preprocessing because incomplete data can reduce model accuracy and lead to biased results. Missing values may occur due to data entry errors, system failures, or unavailable information. The process of handling missing values generally includes the following steps:

First, identify and analyze the missing data to understand how much data is missing and whether there is a pattern (random or systematic).

Second, decide on an appropriate strategy based on the amount and importance of the missing data. If only a small portion of data is missing, rows or columns may be removed. However, if the missing data is significant, imputation techniques are preferred to preserve information.

Third, apply a suitable imputation method and evaluate whether the dataset remains consistent and meaningful after treatment.

Imputation techniques:

For numerical data, a common technique is Mean Imputation. In this method, missing numerical values are replaced with the mean (average) of the available values in that column. This helps maintain the dataset size and is simple to implement.

For categorical data, a common technique is Mode Imputation. Here, missing values are replaced with the most frequently occurring category in that column. This works well when one category appears more often than others.

```python
#Question 8: Write a Python script to:
#● Generate 1000 samples from a normal distribution.
#● Introduce 50 random missing values.
#● Fill missing values with the column mean.
#● Plot a histogram before and after imputation.
#(Include your Python code and output in the code box below.)

# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Generate 1000 samples from a normal distribution
np.random.seed(42)
data = np.random.normal(loc=0, scale=1, size=1000)

# Create DataFrame
df = pd.DataFrame({'Values': data})

# Step 2: Introduce 50 random missing values
missing_indices = np.random.choice(df.index, 50, replace=False)
df.loc[missing_indices, 'Values'] = np.nan

# Plot histogram BEFORE imputation
plt.figure()
plt.hist(df['Values'], bins=30)
plt.title("Histogram Before Imputation")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

# Step 3: Fill missing values with column mean
mean_value = df['Values'].mean()
df['Values'].fillna(mean_value, inplace=True)

# Plot histogram AFTER imputation
plt.figure()
plt.hist(df['Values'], bins=30)
plt.title("Histogram After Imputation")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

# Print confirmation
print("Number of missing values after imputation:", df['Values'].isna().sum())
print("Mean value used for imputation:", mean_value)
```

Step-by-Step Data Preparation Plan

Before training a machine learning model, I would follow these steps:

1. Inspect the Dataset

Start by checking data types, missing values, class balance, and basic statistics to understand the problem.

2. Handle Missing Ages

Since age is numerical, I would fill missing values using mean or median imputation (median is safer if data is skewed).

3. Treat Outliers in Transaction Amount

Outliers can distort model learning. I would use IQR capping (winsorization) to limit extreme values rather than deleting rows, preserving important fraud cases.

4. Encode Categorical Variables

Payment method is categorical, so I would apply one-hot encoding using pandas to convert it into numerical features.

5. Address Class Imbalance

Fraud datasets are usually highly imbalanced. I would use SMOTE (Synthetic Minority Over-sampling Technique) to generate synthetic fraud samples so the model learns both classes properly.

6. Scale Numerical Features

Finally, I would apply Min-Max scaling or Standardization to ensure numerical features are on similar ranges.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from imblearn.over_sampling import SMOTE

# Sample dataset
data = {
    'Age': [25, np.nan, 45, 30, np.nan],
    'Amount': [100, 5000, 120, 80, 10000],
    'Payment': ['Card', 'Cash', 'Card', 'UPI', 'Cash'],
    'Fraud': [0, 0, 0, 1, 0]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# 1. Handle missing ages (median)
df['Age'].fillna(df['Age'].median(), inplace=True)

# 2. Handle outliers using IQR capping
Q1 = df['Amount'].quantile(0.25)
Q3 = df['Amount'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df['Amount'] = np.clip(df['Amount'], lower, upper)

# 3. One-hot encode payment method
df = pd.get_dummies(df, columns=['Payment'])

# 4. Handle class imbalance using SMOTE
X = df.drop('Fraud', axis=1)
y = df['Fraud']

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

# 5. Scale numerical columns
scaler = MinMaxScaler()
X_resampled[['Age','Amount']] = scaler.fit_transform(X_resampled[['Age','Amount']])

print("\nProcessed Features:")
print(X_resampled)

print("\nTarget Distribution After SMOTE:")
print(y_resampled.value_counts())
```
